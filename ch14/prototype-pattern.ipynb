# 프로토타입(Prototype) 패턴
객체를 새로 생성하는 비용이 너무 클 때, 이미 만들어진 객체를 **복사(Copy)** 해서 새로운 객체를 만드는 방식

## 장점
1. **클래스 의존성 감소**: 구체적인 클래스 이름을 몰라도 객체를 복제할 수 있습니다.
2. **다양한 상태의 객체 관리**: 초기화된 상태가 각기 다른 객체들을 미리 만들어두고, 필요할 때 복사해서 즉시 사용할 수 있습니다.
3. **생성 비용 절감**: 객체 생성 과정이 복잡하거나, 리소스를 많이 먹는 경우 복사하는 게 훨씬 빠릅니다.

## 주요 구성 요소
- Prototype (원형): 자기 자신을 복사할 수 있는 인터페이스를 정의합니다.
- Concrete Prototype (실제 원형): 실제로 복사되는 구체적인 클래스입니다.
- Client (사용자): 원형 객체에 복사를 요청하여 새로운 객체를 얻습니다.

```kotlin
// Prototype 역할을 하는 데이터 클래스
data class Monster(
    val name: String,
    val health: Int,
    val skills: List<String>
)

// 1. 원형(Prototype) 생성 - 생성 시 복잡한 로직이 들어갔다고 가정
val basicZombie = Monster("좀비", 100, listOf("물기", "할퀴기"))

// 2. 원형을 복제하여 새로운 객체 생성 (Prototype 패턴의 핵심)
val fastZombie = basicZombie.copy(name = "빠른 좀비", health = 50)
val bossZombie = basicZombie.copy(name = "대장 좀비", health = 500)

println(basicZombie)
println(fastZombie)
println(bossZombie)
```

## 단점
- **복사 오류 (깊은 복사 문제)**: 객체 안에 다른 객체가 들어있는 경우, 데이터가 아닌 '주소값'만 복사되어 원본과 복사본이 의도치 않게 연결될 수 있습니다.
- **순환 참조의 늪**: A가 B를 참조하고 B가 다시 A를 참조하는 구조라면, 복사 로직이 무한 루프에 빠지거나 매우 복잡해집니다.
- **생성자 무시**: 생성자를 거치지 않고 객체를 복제하는 경우가 많아, 초기화 로직이나 보안 검사를 건너뛰게 될 위험이 있습니다.
